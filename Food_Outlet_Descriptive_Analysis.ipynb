```python
df = df.dropna()
```

```python
df.isnull().sum()
```

```python
city_count = df["city"].value_counts()

print(city_count)
```

```python
cuisine_count = df["cuisines"].value_counts().head(10)

print(cuisine_count)
```

```python
top_restaurants = df.sort_values(by="aggregate_rating", ascending=False)[["name", "city", "aggregate_rating"]].head(10)

print(top_restaurants)
```

```python
average_rating = df["aggregate_rating"].mean()

print("Average Rating:", round(average_rating, 2))
```

```python
delivery_count = df["delivery"].value_counts()

print(delivery_count)
```

```python
city_count = df["city"].value_counts().head(10)

plt.figure(figsize=(10,5))
city_count.plot(kind="bar")

plt.title("Top 10 Cities with the Most Restaurants")
plt.xlabel("City")
plt.ylabel("Number of Restaurants")

plt.xticks(rotation=45)
plt.show()
```

```python
cuisine_count = df["cuisines"].value_counts().head(10)

plt.figure(figsize=(10,5))
cuisine_count.plot(kind="bar")

plt.title("Top 10 Most Popular Cuisines")
plt.xlabel("Cuisine")
plt.ylabel("Number of Restaurants")

plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["aggregate_rating"], bins=10)

plt.title("Distribution of Restaurant Ratings")
plt.xlabel("Aggregate Rating")
plt.ylabel("Number of Restaurants")

plt.show()
```

## Conclusion

The analysis provided useful insights into restaurant data, including ratings, cuisines, cities, and delivery services. These findings can help businesses understand customer preferences and make better decisions.
