```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Price (₹)": [20, 25, 30, 35, 40],
    "Bottles Sold": [120, 110, 90, 80, 70]
}

df = pd.DataFrame(data)
df
```

```python
df
```

```python
df["Revenue (₹)"] = df["Price (₹)"] * df["Bottles Sold"]
```

```python
df
```

```python
highest_revenue = df["Revenue (₹)"].max()
print("Highest Revenue:", highest_revenue)
```

```python
best_prices = df[df["Revenue (₹)"] == highest_revenue]["Price (₹)"].tolist()

print("Price(s) with the highest revenue:", best_prices)
```

```python
plt.figure(figsize=(7,5))

plt.bar(df["Price (₹)"], df["Revenue (₹)"])

plt.title("Revenue Earned at Different Selling Prices")
plt.xlabel("Selling Price (₹)")
plt.ylabel("Revenue (₹)")

plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```

## Conclusion

The analysis indicates that the maximum revenue earned is ₹2800, which occurs at both ₹35 and ₹40. Since the ₹35 option sells more bottles while earning the same revenue as ₹40, it is the more practical selling price. Choosing ₹35 helps maintain stronger customer demand without reducing total revenue.
